```python
#import the necessary libraries
import os
import glob
import json
import re
import dask.dataframe as dd
import pandas as pd
import json
```

```python
# Define the directory containing the JSON files
directory_path = '/var/lib/private/simonjaned87761/team-04/data/opendata_api/parsed_protocols/'

# Directory to save the modified JSON file
save_directory = '/var/lib/private/simonjaned87761/team-04/data/opendata_api/'
# Ensure the save directory exists
os.makedirs(save_directory, exist_ok=True)
```

```python
# Define the regex patterns to match the specified patterns
pattern1 = r'((?:Dr\.\s)?(?:\w+(?:-\w+)?\s)(?:\w+\.\s)?(?:von\s)?\w+(?:-\w+)?)\s+\[(CDU\/CSU|CSU|CDU|GRÜNE|FDP|AfD|SPD|KPD|BP|DP|WAV|Z|DIE LINKE|fraktionslos)\]:\s([\s\S]*?)[-–—\)]'
pattern2 = r'[-–—\s\(]Zuruf .*?(CDU\/CSU|CSU|CDU|GRÜNE|FDP|AfD|SPD|KPD|BP|DP|WAV|Z|DIE LINKE|fraktionslos):\s([\s\S]*?)[-–—\)]'

# Combine the patterns using an OR
combined_pattern = f'{pattern1}|{pattern2}'

# Define the output file path
output_file_path = os.path.join(save_directory, 'cleaner_modified_texts_with_interruption.json')

# Initialize the output file
with open(output_file_path, 'w') as outfile:
    json.dump([], outfile)

# Get a list of all JSON files in the directory
json_files = glob.glob(os.path.join(directory_path, '*.json'))

# Process each JSON file and append results to the output file incrementally
for json_file_path in json_files:
    print(f'Processing file: {json_file_path}')  # Debug print
    try:
        # Open and load the JSON file
        with open(json_file_path, 'r') as file:
            data = json.load(file)
        
        # Check if data is loaded correctly
        if not data:
            print(f'No data found in {json_file_path}')
            continue

        # List to store modified texts for the current file
        text_list = []

        # Iterate over each speech in the data
        for speech in data:
            text = speech.get('text', '')
            # Replace matches with <interruption>
            modified_text = re.sub(combined_pattern, '<interruption>', text)
            text_list.append({
                'original_text': speech.get('text', ''),
                'modified_text': modified_text
            })
        
        # Append the modified text list to the output file
        with open(output_file_path, 'r+') as outfile:
            # Load existing data
            existing_data = json.load(outfile)
            # Append new data
            existing_data.extend(text_list)
            # Write back to file
            outfile.seek(0)
            json.dump(existing_data, outfile, indent=4)
    except json.JSONDecodeError as e:
        print(f'Error decoding JSON from file {json_file_path}: {e}')
    except Exception as e:
        print(f'Unexpected error processing file {json_file_path}: {e}')

print(f'Modified texts saved to {output_file_path}')
```
