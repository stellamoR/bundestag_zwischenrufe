```python
import json
import re
import pandas as pd
import matplotlib.pyplot as plt

# Pfad zur JSON-Datei
file_path = '/var/lib/private/simonjaned87761/team-04/data/opendata_api/modified_texts_with_interruption.json'

# JSON-Datei laden
with open(file_path, 'r', encoding='utf-8') as file:
    data = json.load(file)

# Funktion zur Extraktion der Parteizugehörigkeit aus den Einwürfen
def extract_parties(text):
    # Suche nach Mustern wie "(Beifall bei der AfD)" oder "Zurufe von der CDU/CSU"
    party_patterns = re.findall(r'\((Beifall bei|Zurufe von) der ([A-Za-z/]+)\)', text)
    return [party for _, party in party_patterns]

# Aggregation der Einwürfe nach Partei
party_counts = {}

for entry in data:
    modified_text = entry['modified_text']
    parties = extract_parties(modified_text)
    for party in parties:
        if party in party_counts:
            party_counts[party] += 1
        else:
            party_counts[party] = 1

# Umwandlung in DataFrame
df = pd.DataFrame(list(party_counts.items()), columns=['Partei', 'Anzahl'])

# Visualisierung
plt.figure(figsize=(10, 6))
plt.bar(df['Partei'], df['Anzahl'], color=['blue', 'green', 'red', 'yellow', 'black', 'purple'])  # Farben anpassen
plt.xlabel('Partei')
plt.ylabel('Anzahl der Einwürfe')
plt.title('Verteilung der Einwürfe nach Partei')
plt.xticks(rotation=45)
plt.show()
```

```python
#Implementierung eines Sequenzmodells zur Visualisierung von Wahrscheinlichkeiten für Einwürfe 
#entlang des Textes, um Muster zu erkennen

import json
import nltk
from nltk.tokenize import sent_tokenize
from transformers import BertTokenizer, BertForSequenceClassification
import torch
import matplotlib.pyplot as plt
import re

# Sicherstellen, dass die notwendigen NLTK-Ressourcen heruntergeladen sind
nltk.download('punkt')

# Pfad zur JSON-Datei
file_path = '/var/lib/private/simonjaned87761/team-04/data/opendata_api/modified_texts_with_interruption.json'

# JSON-Datei laden
with open(file_path, 'r', encoding='utf-8') as file:
    data = json.load(file)

# Funktion zur Tokenisierung und n-Gramme Extraktion
def extract_ngrams(text, num):
    stopwords = set(['dass', 'der', 'die', 'bei', 'in', 'das', 'ist', ',', '(', ')', '.', ';', '"'])
    text = re.sub(r'[.,();"]', '', text)
    tokens = [word for word in nltk.word_tokenize(text) if word.lower() not in stopwords]
    n_grams = ngrams(tokens, num)
    return [' '.join(grams) for grams in n_grams]

# Laden des Tokenizers und Modells
tokenizer = BertTokenizer.from_pretrained('bert-base-german-cased')
model = BertForSequenceClassification.from_pretrained('bert-base-german-cased')
model.eval()

# Funktion zur Vorhersage
def predict_interruption(text):
    inputs = tokenizer(text, return_tensors='pt', truncation=True, padding=True)
    with torch.no_grad():
        outputs = model(**inputs)
    probabilities = torch.nn.functional.softmax(outputs.logits, dim=-1)
    return probabilities[0][1].item()  # Wahrscheinlichkeit für Klasse "Einwurf"

# Gesamter Prozess der Textsegmentierung, Vorhersage und Visualisierung
all_predictions = []

for entry in data:
    modified_text = entry['modified_text']
    segments = sent_tokenize(modified_text)  # Segmentierung des Textes in Sätze
    
    # Vorhersage für jedes Segment durchführen
    predictions = []
    for segment in segments:
        if segment.strip():  # Leere Segmente ignorieren
            prob = predict_interruption(segment)
            predictions.append((segment, prob))
    
    all_predictions.extend(predictions)

# Ergebnisse anzeigen
for sentence, prob in all_predictions:
    print(f"Sentence: {sentence}\nProbability of Interruption: {prob:.2f}\n")

# Visualisierung der Wahrscheinlichkeiten
probs = [prob for _, prob in all_predictions]
plt.figure(figsize=(10, 6))
plt.plot(probs, marker='o')
plt.xlabel('Segment Index')
plt.ylabel('Probability of Interruption')
plt.title('Probability of Interruption for Each Segment')
plt.grid(True)
plt.show()
```

```python
#Zeitreihenanalyse:
#Analyse, ob sich die Häufigkeit und die Partei der Einwürfe über die Jahre verändert haben

import json
import pandas as pd
import matplotlib.pyplot as plt
from datetime import datetime
import re

# Pfad zur JSON-Datei
file_path = '/var/lib/private/simonjaned87761/team-04/data/opendata_api/modified_texts_with_interruption.json'

# JSON-Datei laden
with open(file_path, 'r', encoding='utf-8') as file:
    data = json.load(file)

# Funktion zur Extraktion der Einwürfe und Kontextinformationen
def extract_context_info(data):
    interruptions = []
    for entry in data:
        modified_text = entry['modified_text']
        speaker = entry.get('speaker', 'unknown')
        party = entry.get('party', 'unknown')
        date = entry.get('date', 'unknown')
        segments = modified_text.split('<interruption>')
        
        for i in range(len(segments) - 1):  # Segmente vor dem Token <interruption>
            interruptions.append({'date': date, 'speaker': speaker, 'party': party, 'text': segments[i].strip()})
    return interruptions

# Kontextinformationen extrahieren
context_data = extract_context_info(data)

# In DataFrame umwandeln
df_context = pd.DataFrame(context_data)

# Datum in datetime-Format konvertieren
df_context['date'] = pd.to_datetime(df_context['date'])

# Jahr extrahieren
df_context['year'] = df_context['date'].dt.year

# Datenaggregation: Anzahl der Einwürfe pro Jahr und Partei
yearly_party_counts = df_context.groupby(['year', 'party']).size().unstack(fill_value=0)

# Visualisierung: Anzahl der Einwürfe pro Jahr und Partei
plt.figure(figsize=(12, 8))
for party in yearly_party_counts.columns:
    plt.plot(yearly_party_counts.index, yearly_party_counts[party], marker='o', label=party)

plt.xlabel('Jahr')
plt.ylabel('Anzahl der Einwürfe')
plt.title('Häufigkeit der Einwürfe pro Jahr und Partei')
plt.legend(title='Partei')
plt.grid(True)
plt.show()
```
