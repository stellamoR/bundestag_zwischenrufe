```python
import requests
import pandas as pd
import json
```

```python
def dict_to_html_params(data):
    html_params = ""
    for key, value in data.items():
        if value != None:
            html_params += f"&{key}={value}"
    return html_params.strip()
```

```python
url_static= "https://search.dip.bundestag.de/api/v1/plenarprotokoll-text?format=json&apikey=rgsaY4U.oZRQKUHdJhF9qguHMkwCGIoLaqEcaHjYLF"

params = {
        "f.datum.start": "2017-01-01", # Change here for different Dates
        "f.datum.end": "2024-05-02",
        "f.zuordnung": "BT",
        "cursor" : None
    }

doc_count = 0
while(True):
    
    api_url = url_static + dict_to_html_params(params)
    print(api_url)

    response = requests.get(api_url)
    print(response)

    if response.status_code == 200:

        with open('./data/opendata_api/requestlibrary.json', 'w', encoding='utf-8') as json_file:
            json.dump(response.json(),json_file, indent=4)

        doc_list = response.json()['documents']
        doc_count += len(doc_list)
        for doc in doc_list:
            path = f'./data/opendata_api/{doc['wahlperiode']}_{doc['dokumentnummer'].split(r'/')[1].zfill(3)}_{doc['datum']}.json'
            print(path)
            with open(path, 'w', encoding='utf-8') as json_file:
                json.dump(doc,json_file, indent=4)
    else:
         print("Error in call. Aborting")
         break
    
    if 'cursor' not in response.json():
        print(f"Retrieved all {doc_list}Documents")
        break 
    elif response.json()['cursor']== params['cursor']:
        print(f"Retrieved all {doc_count} Documents")
        break
    else:
        params['cursor'] = response.json()['cursor']

    
```
