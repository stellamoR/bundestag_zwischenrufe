```python
import xml.etree.ElementTree as ET
```

```python
tree = ET.parse('./data/MDB_STAMMDATEN.XML')
root = tree.getroot()
bt_list = []
# dict for saving in json
for mdb in root.findall('MDB'):
    dict = {
        'last_name':mdb.find('.//NACHNAME').text,
        'first_name':mdb.find('.//VORNAME').text,
        'anrede':mdb.find('.//ANREDE_TITEL').text,
        'party':mdb.find('.//PARTEI_KURZ').text,
        'election_period':mdb.find('.//WP').text
    }
    if dict['party'] == 'BÜNDNIS 90/DIE GRÜNEN':
        dict['party'] = "GRÜNE"
    bt_list.append(dict)

bt_tuples = [(f"{adbt['first_name']} {adbt['last_name']}", adbt['party'], adbt['election_period']) for adbt in bt_list]
bt_tuples = sorted(bt_tuples, key = lambda x: int(x[2]), reverse = True)
bt_tuples[:5]

```

```python
def find_duplicates_by_name(list_of_tuples):
    name_map = {}
    duplicates = []
    
    for tup in list_of_tuples:
        name = tup[0]  # Assuming name is the first element in each tuple
        if name in name_map:
            name_map[name].append(tup)
            if len(name_map[name]) == 2:
                duplicates.extend(name_map[name])
        else:
            name_map[name] = [tup]
    
    return duplicates

find_duplicates_by_name(bt_tuples)
```

#### Conclusion
Not that many duplicates. After all, there are 4379 members in that list.
Since we only have to look up the party for acting ministers, not taking this case into account will likely not interfere with our goals.

```python
def find_party_by_name(name, tuple_list):
    
    if name.startswith('Dr.'):
        name = name[4:]
    for item in tuple_list:
        if name in item[0]:
            return item[1]
    return "Party not found"
# because the tuple list is sorted in descending order by BT-Period it will return newer members of BT first
# (less likely to make error since we mainly look at new data anyways)

print(find_party_by_name('Dr. Robert Habeck', bt_tuples))
```
