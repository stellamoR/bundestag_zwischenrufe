# Bühne Bundestag

## Aufbau

Unser Projekt ist in 5 Teile gegliedert.
- Im Ordner `I_data_acquisition` sind die Notebooks, die für die Datensammlung geschrieben wurden.  
    - Das erste Notebook [01_get_protocol_data](./I_data_acquisition/01_get_protocol_data.ipynb) lädt die Protokolle über die Opendata-Api des Bundestags. (Robin Sternberg)
    - Im zweiten Notebook [02_get_abgeordneten_data](./I_data_acquisition/02_get_abgeordneten_data.ipynb) werden Stammdaten aller bisherigen Bundestagsabgeordneten ausgelesen und für das Text-Parsing aufbereitet (Robin Sternberg)
    - Im dritten Notebook [03_text_parsing](./I_data_acquisition/03_text_parsing.ipynb) werden die in Plaintext vorliegenden Protokolle in Json-Dateien umgewandelt, in denen Daten zu den Rednern, den Zwischenrufen und Beifall leicht durchsuchbar sind. (Robin Sternberg)
- Im Ordner `II_statistical_analysis` gehen wir auf den Aufbau unserer Daten ein. Leitfragen sind: Welche Begriffe waren wann relevant? Wer ruft wem wie oft hinein?
- Im Ordner `III_sentiment_analysis` wird auf Grundlage von eigens gelabelten Daten ein Sentiment-Klassifikator trainiert, der dann auf den restlichen, ungelabelten Daten zur Analyse genutzt wird. 
- Im Ordner `IV_interruption_prediction` werden Modelle zur Zwischenruf-Vorhersage getestet.
    - Das erste (nullte) Notebook [00_Interruption_Tokenization.ipynb](./IV_interruption_prediction
/Every_Model_As_A_Different_Approach_And_Evaluation/00_Interruption_Tokenization.ipynb) fügt an den Stellen, wo Unterbrechungen in Reden sind, <interruption> Token ein. (Edward Simonjan)
- Im Ordner `V_interruption_generation` werden Zwischenrufe auf Grundlage des vorangegangenen Texts neu generiert.
