# Bühne Bundestag

## Aufbau

Unser Projekt ist in 5 Teile gegliedert.
- Im Ordner `I_data_acquisition` sind die Notebooks, die für die Datensammlung geschrieben wurden.  
    - Das erste Notebook [01_get_protocol_data](./I_data_acquisition/01_get_protocol_data.ipynb) lädt die Protokolle über die Opendata-Api des Bundestags. (Robin Sternberg)
    - Im zweiten Notebook [02_get_abgeordneten_data](./I_data_acquisition/02_get_abgeordneten_data.ipynb) werden Stammdaten aller bisherigen Bundestagsabgeordneten ausgelesen und für das Text-Parsing aufbereitet (Robin Sternberg)
    - Im dritten Notebook [03_text_parsing](./I_data_acquisition/03_text_parsing.ipynb) werden die in Plaintext vorliegenden Protokolle in Json-Dateien umgewandelt, in denen Daten zu den Rednern, den Zwischenrufen und Beifall leicht durchsuchbar sind. (Robin Sternberg)
- Im Ordner `II_statistical_analysis` gehen wir auf den Aufbau unserer Daten ein. Leitfragen sind: Welche Begriffe waren wann relevant? Wer ruft wem wie oft hinein?
    - Im ersten Notebook [01_WordClouds_Speeches.ipynb](./II_statistical_analysis/01_WordClouds_Speeches.ipynb) werden die häufigsten Bigramme aller Protokolle in Wordclouds dargestellt. (Edward Simonjan)
    - Im zweiten Notebook [02_Quantity_of_Words_Graph.ipynb](./II_statistical_analysis/02_Quantity_of_Words_Graph.ipynb) wird ein interaktiver Graph erstellt, der die Häufigkeit der eingegebenen Wörter pro Jahr visualisiert. (Edward Simonjan)
- Im Ordner `III_sentiment_analysis` wird auf Grundlage von eigens gelabelten Daten ein Sentiment-Klassifikator trainiert, der dann auf den restlichen, ungelabelten Daten zur Analyse genutzt wird. 
- Im Ordner `IV_interruption_prediction` werden Modelle zur Zwischenruf-Vorhersage getestet.
    - alle Modelle befinden sich im Ordner [Every_Model_As_A_Different_Approach_And_Evaluation](./Every_Model_As_A_Different_Approach_And_Evaluation)
    - Das nullte Notebook [00_Interruption_Tokenization.ipynb](./IV_interruption_prediction/Every_Model_As_A_Different_Approach_And_Evaluation/00_Interruption_Tokenization.ipynb) fügt an den Stellen, wo Unterbrechungen in Reden sind, <interruption> Token ein. Alle folgenden Notebooks basieren darauf. (Edward Simonjan)
    - Im ersten Notebook [01_Naive_Bayes.ipynb](./IV_interruption_prediction/Every_Model_As_A_Different_Approach_And_Evaluation/01_Naive_Bayes.ipynb) werden Unterbrechungen mit Hilfe des Naive Bayes **satzweise** vorhergesagt. (Edward Simonjan)
    - Im zweiten Notebook [02_Unfinetuned_BERT.ipynb](./IV_interruption_prediction/Every_Model_As_A_Different_Approach_And_Evaluation/02_Unfinetuned_BERT.ipynb) werden Unterbrechungen **satzweise** mit Hilfe des nicht gefinetunten BERT-Modells vorhergesagt. Dieses verwendet die **Cross-Entropy-Loss** Funktion. (Edward Simonjan)
    - Im dritten Notebook [03_Finetuned_BERT.ipynb](./IV_interruption_prediction/Every_Model_As_A_Different_Approach_And_Evaluation/03_Finetuned_BERT.ipynb) wird BERT finegetuned und unter der Default Verwendung der **Cross-Entropy-Loss** Funktion werden dann **satzweise** Unterbrechungen vorhergesagt. (Edward Simonjan)
    - Im vierten Notebook [04_BERT_Only_Cross_Entropy.ipynb](./IV_interruption_prediction/Every_Model_As_A_Different_Approach_And_Evaluation/04_BERT_Only_Cross_Entropy.ipynb) wird unter der Verwendung der **Cross-Entropy-Loss** Funktion auf **Tokenebende** nach Einwürfen klassifiziert. Der Name dieses Notebooks mag eventuell etwas verwirrend wirken, aber es soll den Unterschied zum nächsten Notebook hervorheben. (Edward Simonjan)
    - Im fünften Notebook [05_BERT_weighted_cross_entropy.ipynb](./IV_interruption_prediction/Every_Model_As_A_Different_Approach_And_Evaluation/05_BERT_weighted_cross_entropy.ipynb) wurde versucht den Task auf **Tokenebene** mit Hilfe der **Weighted-Cross-Entropy-Loss** Funktion besser zu erfüllen. (Edward Simonjan)
    - Das sechste Notebook [06_My_RoBERTa_Modell.ipynb](./IV_interruption_prediction/Every_Model_As_A_Different_Approach_And_Evaluation/06_My_RoBERTa_Modell.ipynb) benutzt das leistungsstärkere BERT-Modell namens RoBERTa. Außerden wird hier auch die **Cross-Entropy-Loss** Funktion verwendet und auf **Tokenebene** klassifiziert. (Edward Simonjan)
    - Im siebten Notebook [07_Fewshot-Prompting.ipynb](./IV_interruption_prediction/Every_Model_As_A_Different_Approach_And_Evaluation/07_Fewshot-Prompting.ipynb) wird mit Fewshot-Prompting versucht die Vorhersagen der Einwürfe zu machen. (Edward Simonjan)
- Im Ordner `V_interruption_generation` werden Zwischenrufe auf Grundlage des vorangegangenen Texts neu generiert.
