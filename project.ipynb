# Bühne Bundestag

## Aufbau

Unser Projekt ist in 5 Teile gegliedert.
- Im Ordner [01_data_acquisition](./01_data_acquisition/) sind die Notebooks, die für die Datensammlung geschrieben wurden.  
    - Das erste Notebook [01_get_data](./01_data_acquisition/01_get_data.ipynb) lädt die Protokolle über die Opendata-Api des Bundestags. (Robin Sternberg)
    - Im zweiten Notebook [02_get_abgeordneten_data](./01_data_acquisition/02_get_abgeordneten_data.ipynb) werden die Stammdaten aller bisherigen Bundestagsabgeordneten ausgelesen und für das Text-Parsing aufbereitet (Robin Sternberg)
    - Im dritten Notebook [03_text_parsing](./01_data_acquisition/03_text_parsing.ipynb) werden die in Plaintext vorliegenden Protokolle in Json-Dateien umgewandelt, in denen Daten zu den Rednern, den Zwischenrufen und Beifall leicht durchsuchbar sind. (Robin Sternberg)
- Im Ordner [02_statistical_analysis](./02_statistical_analysis/) gehen wir auf den Aufbau unserer Daten ein. Leitfragen sind: Welche Begriffe waren wann relevant? Wer ruft wem wie oft hinein?
- Im Ordner [03_sentiment_analysis](./03_sentiment_analysis/) wird auf Grundlage von eigens gelabelten Daten ein Sentiment-Klassifikator trainiert, der dann auf den restlichen, ungelabelten Daten zur Analyse genutzt wird. 
- Im Ordner [04_interruption_prediction](./04_interruption_prediction//) werden Modelle zur Zwischenruf-Vorhersage getestet.
- Im Ordner [05_interruption_generation](./04_interruption_prediction//) werden Zwischenrufe auf Grundlage des vorangegangenen Texts neu generiert.
